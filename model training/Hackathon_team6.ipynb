```python
from ultralytics import YOLO

model = YOLO("ultralytics/yolov8n.pt")

# Train the model
model.train(
    data="model.yaml",  # Path to dataset config
    epochs=50,         # Number of epochs
    imgsz=640,         # Image size
    batch=16,          # Adjust batch size based on GPU memory
    device="cuda",
    patience=5
)
```

```python
results = model.val()
print(results)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the training logs
log_path = "runs/detect/train3/results.csv"  # Adjust path if needed
df = pd.read_csv(log_path)

# Extract loss values
epochs = df["epoch"]
train_loss = df["train/box_loss"]
val_loss = df["val/box_loss"]

# Plot training & validation loss
plt.figure(figsize=(8, 5))
plt.plot(epochs, train_loss, 'bo-', label='Train Loss')
plt.plot(epochs, val_loss, 'r*-', label='Validation Loss')
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training & Validation Loss")
plt.legend()
plt.show()
```
